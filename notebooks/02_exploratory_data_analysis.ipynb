```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

from src.eda import (
    set_plotting_style,
    get_figures_dir,
    plot_overall_sales_trend,
    plot_holiday_vs_regular_sales,
    plot_store_performance_tiers,
    plot_correlation_heatmap,
    plot_monthly_seasonality,
    generate_eda_report
)

set_plotting_style()
figures_dir = get_figures_dir()
clean_data_path = project_root / 'data' / 'processed' / 'clean_sales.csv'
df = pd.read_csv(clean_data_path)
df['Date'] = pd.to_datetime(df['Date'])
print(f'Clean data loaded: {df.shape[0]:,} records across {df["Store"].nunique()} stores.')
```

## 1. Overall Retail Sales Trend & Seasonal Spikes
Examines total sales volume aggregated across all 45 stores over time.

```python
import matplotlib.image as mpimg

fig1_path = plot_overall_sales_trend(df, figures_dir)
img1 = mpimg.imread(str(fig1_path))
plt.figure(figsize=(14, 6), dpi=100)
plt.imshow(img1)
plt.axis('off')
plt.show()
```

### Insight 1:
- Sales follow a stable baseline of ~$45M to $50M per week nationwide.
- Clear annual peaks occur in late November (Thanksgiving / Black Friday) and late December (Christmas rush), surging sales up to ~$80M per week (a ~60% spike).
- Smaller secondary spikes appear around Super Bowl (early February) and Labor Day (early September).

## 2. Holiday vs. Non-Holiday Demand Uplift
Compares sales distribution on designated holiday weeks versus standard weeks.

```python
fig2_path = plot_holiday_vs_regular_sales(df, figures_dir)
img2 = mpimg.imread(str(fig2_path))
plt.figure(figsize=(14, 5.5), dpi=100)
plt.imshow(img2)
plt.axis('off')
plt.show()
```

```python
eda_metrics = generate_eda_report(df)
print(f'Mean Regular Weekly Sales: ${eda_metrics["mean_regular_sales"]:,.2f}')
print(f'Mean Holiday Weekly Sales: ${eda_metrics["mean_holiday_sales"]:,.2f}')
print(f'Overall Holiday Uplift:    +{eda_metrics["holiday_sales_uplift_pct"]:.2f}%')
```

### Insight 2:
- Average weekly store sales during holidays are **$1,122,887.89** versus **$1,041,256.38** on regular weeks, representing an average demand uplift of **+7.84%**.
- Specific holiday weeks like Black Friday exhibit over **+35% to +60%** store-level spikes.

## 3. Store Performance Heterogeneity
Ranks all 45 outlets by total accumulated revenue.

```python
fig3_path = plot_store_performance_tiers(df, figures_dir)
img3 = mpimg.imread(str(fig3_path))
plt.figure(figsize=(15, 6), dpi=100)
plt.imshow(img3)
plt.axis('off')
plt.show()
```

### Insight 3:
- Massive variance exists across stores. Top-performing store 20 generated **$301.40M**, whereas store 33 generated **$37.16M** (an 8.1x difference!).
- This requires models to capture store-level base volume, and indicates inventory allocation cannot be one-size-fits-all.

## 4. Feature Correlation Matrix
Evaluates linear relationships between sales and macroeconomic / environmental regressors.

```python
fig4_path = plot_correlation_heatmap(df, figures_dir)
img4 = mpimg.imread(str(fig4_path))
plt.figure(figsize=(9, 7.5), dpi=100)
plt.imshow(img4)
plt.axis('off')
plt.show()
```

### Insight 4:
- Environmental variables (`CPI`, `Unemployment`, `Temperature`, `Fuel_Price`) show weak direct linear correlation with weekly sales.
- Store identity and calendar seasonality (`Month`, `WeekOfYear`, `Holiday_Flag`, `Is_PreChristmas`) are the dominant drivers of demand.

## 5. Monthly Seasonality Analysis

```python
fig5_path = plot_monthly_seasonality(df, figures_dir)
img5 = mpimg.imread(str(fig5_path))
plt.figure(figsize=(10, 5.5), dpi=100)
plt.imshow(img5)
plt.axis('off')
plt.show()
```
